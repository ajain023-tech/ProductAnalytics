# Phase 5 — Feature Engineering

## Objective
Create a small, interpretable dataset for a future Logistic Regression baseline predicting `d30_retained`. This notebook creates features and validates them only; it does not train or evaluate a model.

## Feature definitions and leakage checks

The five model input columns are `experiment_treatment`, `onboarding_completed`, `trial_started`, `paid_conversion`, and `d1_retained`. Neither `early_engagement_score` nor `journey_progress_score` is included in the model feature matrix. `d7_retained` is excluded.

`d30_retained` is the target only, and `user_id` is retained only as a unique row identifier. Outcome timestamps are unavailable, so verify predictor timing against the prediction cutoff. No model is trained in this notebook.


```python
from pathlib import Path
import sys

project_root = Path.cwd()
if not (project_root / "python" / "feature_engineering.py").is_file():
    project_root = Path.cwd().parent
sys.path.insert(0, str(project_root / "python"))

from feature_engineering import (
    FEATURE_COLUMNS,
    TARGET_COLUMN,
    create_feature_summary,
    create_model_features,
    load_cleaned_data,
    save_outputs,
    validate_model_features,
)
```

```python
data = load_cleaned_data(project_root / "outputs" / "cleaned_data" / "cleaned_ab_test_dataset.csv")
print("Cleaned dataset shape:", data.shape)
```

## Feature matrix

The numeric matrix contains only the five documented inputs; no derived additive score is included. No timestamps or unavailable user attributes are invented.


```python
model_features = create_model_features(data)
validate_model_features(model_features)
feature_summary = create_feature_summary(model_features)
features_path, summary_path = save_outputs(model_features, feature_summary)
print("Feature dataset:", features_path)
print("Feature summary:", summary_path)
```
