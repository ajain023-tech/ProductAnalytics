# Phase 4 — Statistical Analysis

## Objective
Compare Control and Treatment outcome proportions using the cleaned user-level dataset. These tests describe evidence of differences; they do not by themselves establish causality.

## Hypotheses and test selection

For each outcome, **H0:** Control and Treatment have the same proportion. **H1:** the proportions differ. A two-proportion z-test is appropriate because each outcome is binary (0/1) and we compare proportions in two independent experiment groups. The test is two-sided, uses a pooled proportion under H0, and reports a 95% confidence interval for Treatment minus Control using an unpooled standard error.

The **p-value** is the probability, assuming H0 is true, of seeing a difference at least as extreme as the observed one. It is not the probability that H0 is true. **Alpha = 0.05** is the decision threshold: a p-value below 0.05 is called statistically significant for this test. A **95% confidence interval** is a method that, over repeated comparable samples, would cover the true rate difference in about 95% of intervals.

Onboarding uses all assigned users because no separate onboarding eligibility field exists. Trial, paid, and retention outcomes use their corresponding eligibility flags.

## Calculations

For each group, the rate is successes divided by eligible users. The pooled rate combines successes and denominators from both groups and is used for the null-hypothesis z-test. The confidence interval uses the separate group rates. No adjustment across the six requested tests is applied.
